### 2. Determinar las redes sociales que no están presentes globalmente, operando únicamente en pequeñas áreas geográficas o subconjuntos de países
---

['Instagram' 'Twitter' 'TikTok' 'YouTube' 'Facebook' 'LinkedIn' 'Snapchat' 'LINE' 'KakaoTalk' 'VKontakte' 'WhatsApp' 'WeChat']

**Assistant:**

* **Instagram**

* **TikTok**
* **YouTube**

---

### 🌎 Platforms with More Limited Regional Focus

The platforms you listed that primarily operate in a few specific countries or regions are generally considered to have a more **limited regional focus**:

* **WeChat:** While having a large user base, its primary market and function are heavily concentrated in **China** and Chinese communities abroad.
* **LINE:** Predominantly popular in **Japan**, **Taiwan**, and **Thailand**.
* **KakaoTalk:** Primarily dominant in **South Korea**.
* **VKontakte (VK):** The largest social network in **Russia** and other Russian-speaking countries (Eastern Europe/former Soviet states).

---


---
* Por lo tanto, procedemos a excluir las siguientes redes sociales del análisis:
---

````python
social_media_platforms_to_exclude_for_not_having_global_presence = ["WeChat", "LINE", "KakaoTalk", "VKontakte"]
````

---
* Eliminamos los datos que nos impiden comparar países y redes sociales. 
---

````python
# Eliminar países donde existen prohibiciones de uso de alguna de las redes sociales
mask = ~ origin_data_df["Country"].isin(countries_to_exclude_from_dataset_for_having_prohibitions_on_analyzed_social_media)
temp_df = origin_data_df[mask]

# Eliminar redes sociales sin presencia global
mask = ~ temp_df["Most_Used_Platform"].isin(social_media_platforms_to_exclude_for_not_having_global_presence)
temp_df = temp_df[mask]
                                          
````

---
* Eliminamos la red social "WhatsApp" de entre las redes sociales estudiadas, puesto que esta  red permite mantener llamadas telefónicas que desvirtuarían las métricas de uso de redes sociales puras. Antes de realizar esta eliminación, se ha observado, de hecho, que se trata de la red social donde de media más horas se reportan en la columna "Most_Used_Platform", y que no estamos eliminando un subconjunto muy grande de registros:
---

````python
# Verificamos que el dato "Avg_Daily_Usage_Hours" donde la "Most_Used_Platform" es "WhatsApp", es la mayor de todas; suponemos que la posibilidad de realizar llamadas puede estar adulterando este resultado

mean_daily_usage_hours_grouped_by_most_used_platform = temp_df.groupby("Most_Used_Platform")['Avg_Daily_Usage_Hours'].mean().reset_index()
mean_daily_usage_hours_grouped_by_most_used_platform.columns = ["Most_Used_Platform", "Avg_Daily_Usage_Hours_of_ALL_social_media"]
mean_daily_usage_hours_grouped_by_most_used_platform.sort_values(by="Avg_Daily_Usage_Hours_of_ALL_social_media", ascending=False)
````

````python
numero_de_registros_adicionales_que_se_eliminaran = len(temp_df[temp_df["Most_Used_Platform"] == "WhatsApp"])
print("Número de registros adicionales que se eliminarán al no considerar WhatsApp:", numero_de_registros_adicionales_que_se_eliminaran)
````

````python
# Eliminar registros donde la red social sea WhatsApp en la columna "Most_Used_Platform"
mask = ~ temp_df["Most_Used_Platform"].isin(["WhatsApp"])
clean_dataset_df = temp_df[mask]  
````

---
* Revisar que los datos finales tienen sentido y el número total de registros eliminados
---

````python
# Check the the resulting data looks correct (selected records have  dissapeared):
print("Países únicos:")
print("--------------------------")
print(clean_dataset_df["Country"].unique())
print("________________________________________________________________________________________")

print("Redes sociales únicas:")
print("--------------------------")
print(clean_dataset_df["Most_Used_Platform"].unique())
print("________________________________________________________________________________________")

print("Número inicial de registros:", len(origin_data_df))
print("Número final de registros:", len(clean_dataset_df))
print("NÚMERO DE REGISTROS ELIMINADOS:", len(origin_data_df) - len(clean_dataset_df))
````
