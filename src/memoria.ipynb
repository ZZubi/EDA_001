### 02. Análisis inicial de datos y limpieza de registros que desvirtuarían el objeto de análisis de este EDA
* De cara a poder comparar los datos entre sí en las hipótesis lanzadas inicialmente, se han identificado, cuantificado y eliminado los siguientes registros:
    * **Países** donde existen o han existido prohibiciones parciales o totales de alguna de las redes sociales con las que se trabaja en esta encuesta. No se ha considerado como censura la limitación que algunos países hacen de VoIP para proteger la industria de telecomunicaciones local, y que afectaría a plataformas como WhatsApp
    > Detalle de países eliminados: ['China', 'Iran', 'Russia', 'Turkmenistan', 'India', 'Pakistan', 'Uzbekistan', 'Nepal', 'Bangladesh']
    * **Redes sociales** sin presencia geográfica global; se han eliminado redes con presecia solamente en zonas geográficas muy concretas.
    > Detalle de redes sociales eliminadas: ["WeChat", "LINE", "KakaoTalk", "VKontakte"]
    * **WhatsApp** se ha eliminado también por considerar que no se trata de una red social al uso, sino más bien de una plataforma de comunicación. Se ha verificado que:
        * Los usuarios que reportan esta plataforma como la más utilizada  reportan un uso medio mucho mayor que el siguiente clasificado; suponemos que puede deberse a la utilización para hablar por VoIP, y esto podría distorsionar la comparabilidad de los datos
        * El número adicional de registros eliminados será tan sólo de 29, un 4.1% respecto al número de registros inicial
* De un total de **705 encuestas** realizadas, hemos **eliminado 202**, quedando finalmente la muestra en **503 registros a analizar**

### 03. Hipótesis


#### 3.1 Instagram y Tik Tok son las redes sociales donde, de media, más horas dedican los participantes de la encuesta (detalle [aquí](notebooks/03.01-Hypothesis_testing.ipynb))

Al revisar detenidamente la explicación del campo "Avg_Daily_Usage_Hours", he observado que no es posible verificar la hipótesis cuantitavamente, puesto que el campo mide el total de horas dedicadas en TODAS las redes sociales:

> "Avg_Daily_Usage_Hours: The average number of hours per day the student spends on social media platforms, calculated from self-reported weekday and weekend usage estimates."

Como workaround, estableceremos 2 sub-hipótesis que nos permitan dar por probada nuestra hipótesis principal:

##### Sub-hipótesis 1: A nivel de conteo, Instagram y Tik Tok son las redes sociales que más veces han entrado en la categoría de "Most_Used_Platform" de entre todas las del dataset

![](data/images/01.01-Hypothesis.png)

##### ✅ Como podemos ver en el gráfico, la sub-hipótesis 1 se verifica a la perfección

**Efectivamente, Instagram y Tik Tok son las plataformas que más veces han sido reportadas como la red social más utilizada**


##### Sub-hipótesis 2: Los usuarios que reportan como "Most_Used_Platform" las redes Instagram y Tik Tok, también reportan de media un mayor uso de redes sociales al día

![](data/images/01.02-Hypothesis.png)

##### ❌ Debemos desmentir la Sub-hipótesis 2 

Según los resultados obtenidos, observamos que los usuarios que han reportado TikTok como "Most_Used_Platform" sí que lideran el uso medio de redes sociales. 

Sin embargo, esperábamos que "Instagram" también apareciera en el Top 2, y sin embargo aparece en la cuarta posición. En cualquier caso, cabe destacar que las diferencias de uso medio entre las posiciones 2, 3 y 4 no son demasiado significativas, a nivel relativo están entre el 90.49% y el 95.86% comparando con la agrupación con mayor uso, variando entre ellas menos del 3%

Esto nos lleva a **no poder validar categóricamente esta primera hipótesis**; nos lleva asimismo a tener que reescribir las hipótesis iniciales que estaban basadas en esta primera hipótesis, tal como ha quedado explicado en el fichero [README](../README.md)

#### 3.3 Los usuarios que mayoritariamente usan Instagram y Tik Tok, perciben que el uso de las redes sociales afecta negativamente a su rendimiento académico que aquellos encuestados con preferencia por otras redes sociales (detalle [aquí](notebooks/03.03-Hypothesis_testing.ipynb))

Cabe destacar que Tik Tok y Snapchat están muy cerca entre sí, y que la diferencia es muy grande respecto al resto de plataformas.

#### 3.4 Existe una fuerte correlación **negativa** entre el número de horas dedicados a redes sociales y la puntuación que se otorgan a sí mismos en el apartado de salud mental. Esta hipótesis trata de profundizar en si a nivel autopercibido existe relación entre el elevado uso de redes sociales y una peor salud mental (detalle [aquí](notebooks/03.04-Hypothesis_testing.ipynb))

![](data/images/04_and_05-Hypothesis.png)

##### ✅ Podemos dar por válida la hipótesis, al encontrarse la correlación en el intervalo [-0.6, -1] 

Tal como se puede observar en la matriz de correlación anterior, **existe una fuerte correlación negativa de -0.84** entre las columnas "Avg_Daily_Usage_Hours" y "Mental_Health_Score", lo que nos permite deducir que efectivamente los usuarios encuestados se autoperciben con una PEOR salud mental cuando MAYOR es el número de horas que utilizan redes sociales

#### 3.5 Existe una fuerte correlación **positiva** entre el número de horas dedicados a redes sociales y el número de conflictos que los usuarios tienen con familiares, amigos y compañeros en relación con el uso de estas plataformas (detalle [aquí](notebooks/03.05-Hypothesis_testing.ipynb))

![](data/images/04_and_05-Hypothesis.png)

##### ✅ Podemos dar por válida la hipótesis, al encontrarse la correlación en el intervalo [0.6, 1] 

Tal como se puede observar en la matriz de correlación anterior, **existe una fuerte correlación positiva de 0.82** entre las columnas "Avg_Daily_Usage_Hours" y "Conflicts_Over_Social_Media", lo que nos permite deducir que efectivamente los usuarios encuestados reportan un MAYOR número de conflictos con su entorno cercano cuanto MAYOR es el número de horas que utilizan redes sociales



### 04. Conclusiones

* Al redactar las hipótesis se partía de la idea de que Tik Tok e Instagram eran las redes sociales que mayores problemas de adicción causaban; esta afirmación se ha verificado consistentemente en el caso de Tik Tok, pero no ha sido así en el caso de Instagram
* Ha sido Snapchat la plataforma que ha sustituido a Instagram en este Top 2. Este resultado me ha hecho experimentar en mis propias carnes lo que es un Sesgo de Confirmación, puesto que tracé las hipótesis fijándome en las redes sociales que conozco, habiendo ignorado completamente SnapChat por algo tan circunstancial como que no se usa tanto en España
* El resto de hipótesis sí se han podido verificar, habiendo demostrado que existe efectivamente una alta correlación (negativa y positiva respectivamente) entre las columnas:
    * "Avg_Daily_Usage_Hours" y "Mental_Health_Score"
    * "Avg_Daily_Usage_Hours" y "Conflicts_Over_Social_Media"

* La primera de las anteriores nos permite decir que los usuarios que más horas dedican de media se puntuan a sí mismos con una PEOR salud mental; esto nos transmite la idea de que los usuarios auto-perciben que las redes sociales les afectan negativamente en este apartado

* En el caso de la segunda, observamos que un elevado uso de redes sociales va de la mano con un mayor número de conflictos con familiares, amigos y compañeros.
